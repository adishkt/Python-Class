```python
class Car:
    def __init__(self, name, model, year, speed_limit):
        self.name=name
        self.model=model
        self.year=year
        self.speed_limit=speed_limit

    def is_supercar(self):
        if self.speed_limit>200:
            print("This is a supercar")
        else:
            print("This is not a supercar")
    
    def display(self):
        print(f"Car name is {self.name}")
        print(f"Car model is {self.model}")
        print(f"Car year is {self.year}")
        print(f"Car speed limit is {self.speed_limit}")
        
        
bmw=Car("BMW", "X5", 2020, 250)
print(bmw.name)
print(bmw.model)
bmw.display()
bmw.is_supercar()
```
