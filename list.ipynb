```python
cart=["milk","Bread","Eggs","Butter"]
print("original cart : ",cart)

cart[3]="unsalted butter"
print(cart)
```

```python
cart.append("sugar")
print(cart)

cart.insert(2,"salt")
print(cart)

cart.extend(["oil","salt"])
print(cart)
```

```python
cart.pop()
print(cart)

cart.remove("sugar")
print(cart)

del cart[2]
print(cart)
```

```python
for marks in class_marks:
    for mark in marks:
        print(mark,end=" ")
    print()
```

```python
nums=[23,32,4,33,23,45,32,45,34]
print(nums)

result=[]
for num in nums:
    if num not in result:
        result.append(num)
        
print("Unique value in list",result) 
```
