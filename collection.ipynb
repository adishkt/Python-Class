```python
from collections import namedtuple

Student=namedtuple('Student',['name','age','city'])

s1=Student('Kiran',22,'Bangalore')
s2=Student('Priya',23,'Chennai')

print(s1)
print(s1.name)
print(s1[2])
print(s2.city)
```

```python
from collections import deque

d1=deque()
d1.append(5)
d1.append(10)
print(d1)
d1.appendleft(3)
print(d1)



d1.extend([15,20])
print(d1)
d1.extend([1,2])
print(d1)

d1.pop()
print(d1)
d1.popleft()
print(d1)

```

```python
from collections import Counter,OrderedDict

ob=OrderedDict()
ob['a']=1
ob['b']=2
ob['c']=3
print(ob)
ob.move_to_end('a')
print(ob)

c=Counter('aeroplane')
print(c)
print(c.most_common(2))
print(c.total())

words=["apple","banana","apple","cherry","banana","apple"]
print(Counter(words))
```
