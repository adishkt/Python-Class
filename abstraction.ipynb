```python
from abc import abstractmethod,ABC

class Fruit(ABC):
    
    @abstractmethod
    def shape(self):
        pass
    
    @abstractmethod
    def color(self):
        pass
    
    def category(self):
        print("Edible Fruit")
        
# Fruit() Can't instantiate abstract class Fruit without an implementation for abstract methods 'color', 'shape'

class Mango(Fruit):
    
    # @abstractmethod
    def shape(self):
        print("Oval shape")
        
    # @abstractmethod
    def color(self):
        print("Yellow")

obj=Mango()
obj.shape() 
obj.color()
obj.category()
```
