```python
class GrandParent:
    def gp_method(self):
        print("GrandParent here")



class Parent(GrandParent):
    def get_mobile(self):
        print("Samsung Edge")

class Child(Parent):
    def get_Bike(self):
        print("KTM Duke 390")
        
c1=Child()
c1.get_Bike()
c1.get_mobile()
c1.gp_method()
```
