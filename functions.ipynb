```python
def add_num(a,b):
    print(a+b)
    print(a*b)
    print(a/b)
    print("*"*10)

add_num(10,5)
add_num(20,10)
add_num(30,15)
```

Recursive::::::::::::

```python
def factorial(n):
    if n==0:
        return 1
    else:
        return(n*factorial(n-1))
print(factorial(5))    

```
