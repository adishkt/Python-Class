```python
pwd=input("Enter your password: ")
lower_count=0
upper_count=0
spl_char=0
space=0


if len(pwd)<8:
    print("minimum length of password should be 8")
else:
    if pwd.islower():
        lower_count+=1
    elif pwd.isupper():
        upper_count+=1
    elif pwd.isspace():
        space+=1
    elif not pwd.isalnum() and not pwd.isspace():
        spl_char+=1
if lower_count>0 and upper_count>0 and spl_char>0 and space==0:
    print("Strong password")
else:
    print("Weak password")
```
